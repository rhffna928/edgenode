```python
import re

file_name = "./node_250522.log"

count = 0
delay_times = []
with open(file_name, "r") as f:
    lines = f.readlines()
for line in lines:
    if "nodemqtt_2.py:559" in line:
        count += 1
        match = re.search(r"delay_time:\s([\d.]+)ms",line)

        if match:
            delay_time = float(match.group(1))
            delay_times.append(delay_time)
            print(f"Delay time: {delay_time} ms")
        else:
            print("No delay time found in the log file.")
            
print(f"\n'nodemqtt_2.py:559'가 포함된 줄 수: {count}")
print(f"평균 : {sum(delay_times)/ count} ms")
print(f"최대 : {max(delay_times):.3f} ms")
print(f"최소 : {min(delay_times):.3f} ms")
# if delay_time:
#     print(f"최대 지연 시간: {max(float(delay_time) for delay_time in re.findall(r'([\d.]+)ms', ''.join(lines)))} ms")
```
